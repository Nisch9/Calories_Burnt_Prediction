# Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn import metrics
from sklearn.metrics import r2_score,mean_absolute_error
```

# Data loading and preprocessing

```python
calories = pd.read_csv('../Dataset/calories.csv')
```

```python
calories
```

```python
exercise_data = pd.read_csv('../Dataset/exercise.csv')
```

```python
exercise_data
```

# Combining DF's

```python
calories_data = pd.concat([exercise_data, calories['Calories']], axis=1)
```

```python
calories_data
```

```python
calories_data.info()
```

```python
calories_data.isnull().sum()
```

```python
calories_data.describe()
```

# Visualisation

```python
sns.set()
```

```python
# finding the distribution of "Age" column
sns.distplot(calories_data['Age'])
```

```python
# finding the distribution of "Height" column
sns.distplot(calories_data['Height'])
```

```python
correlation = calories_data.corr()
```

```python
plt.figure(figsize=(10,10))
sns.heatmap(correlation, cbar=True, square=True, fmt='.1f', annot=True, annot_kws={'size':8}, cmap='Reds')
```

```python
calories_data.replace({"Gender":{'male':0,'female':1}}, inplace=True)
```

```python
calories_data
```

# Splitiing into train and test

```python
X = calories_data.drop(columns=['User_ID','Calories','Weight'], axis=1)
Y = calories_data['Calories']
```

```python
X
```

```python
Y
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

# Training

# Linear Regression

```python
model = LinearRegression()
```

```python
model.fit(X_train, Y_train)
```

```python
test_data_prediction = model.predict(X_test)
```

```python
print(test_data_prediction)
```

```python
mae = metrics.mean_absolute_error(Y_test, test_data_prediction)
```

```python
print("Mean Absolute Error = ", mae)
```

```python
print(r2_score(Y_test,test_data_prediction))
```

# Random Forest

```python
model = RandomForestRegressor()
```

```python
model.fit(X_train, Y_train)
```

```python
test_data_prediction = model.predict(X_test)
```

```python
mae = metrics.mean_absolute_error(Y_test, test_data_prediction)
```

```python
print("Mean Absolute Error = ", mae)
```

```python
print(r2_score(Y_test,test_data_prediction))
```

# XGBRegressor

```python
model = XGBRegressor()
```

```python
model.fit(X_train, Y_train)
```
